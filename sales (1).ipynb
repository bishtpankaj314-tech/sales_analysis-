```python
# Extract Year from Order Date
df['Order Year'] = df['Order Date'].dt.year

# Group by year and get the set of unique customers for each year
yearly_customers = df.groupby('Order Year')['Customer ID'].apply(set).to_dict()

years = sorted(yearly_customers.keys())
retention_data = []

# Calculate Year-over-Year Retention
for i in range(len(years) - 1):
    current_year = years[i]
    next_year = years[i+1]

    current_set = yearly_customers[current_year]
    next_set = yearly_customers[next_year]

    # Customers who bought in current year and returned in the next year
    retained_customers = current_set.intersection(next_set)
    retention_rate = (len(retained_customers) / len(current_set)) * 100

    retention_data.append({
        'From Year': current_year,
        'To Year': next_year,
        'Starting Customers': len(current_set),
        'Retained Customers': len(retained_customers),
        'Retention Rate (%)': round(retention_rate, 2)
    })

retention_df = pd.DataFrame(retention_data)
display(retention_df)

# Visualize the Retention Rate
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.barplot(x=retention_df['From Year'].astype(str) + ' -> ' + retention_df['To Year'].astype(str),
            y=retention_df['Retention Rate (%)'],
            hue=retention_df['From Year'].astype(str),
            legend=False,
            palette='viridis')
plt.title('Year-over-Year Customer Retention Rate (%)')
plt.ylabel('Retention Rate (%)')
plt.xlabel('Yearly Cohort Transition')
plt.ylim(0, 100)
for index, row in retention_df.iterrows():
    plt.text(index, row['Retention Rate (%)'] + 2, f"{row['Retention Rate (%)']}%", ha='center')
plt.show()
```

```python
yoy_df = sum_sales_year.reset_index()
yoy_df.columns = ['Year', 'Sales']

yoy_df['Previous Year Sales'] = yoy_df['Sales'].shift(1)
yoy_df['YoY Growth (%)'] = ((yoy_df['Sales'] - yoy_df['Previous Year Sales']) / yoy_df['Previous Year Sales']) * 100

display(yoy_df.round(2))
```
